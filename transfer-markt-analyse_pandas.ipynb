### Bereinigung von Ablöse und Marktwert

Sonderwerte wie ablösefreie Transfers und unbekannte Ablösesummen werden
für die weitere Analyse vereinheitlicht.

```python
bundesliga["Ablöse"].value_counts()
```

```python
bundesliga[bundesliga["Ablöse"].isin(["-", "?", "ablösefrei"])]
```

```python
bundesliga[bundesliga["Marktwert"] == "-"]
```

```python
premier_league["Ablöse"].value_counts()
```

```python
premier_league[premier_league["Ablöse"].isin(["-", "?", "ablösefrei"])]
```

```python
premier_league[premier_league["Marktwert"] == "-"]
```

```python
import numpy as np

bundesliga["Ablöse"] = bundesliga["Ablöse"].replace({
    "ablösefrei": "0",
    "?": np.nan,
    "-": np.nan
})

bundesliga["Marktwert"] = bundesliga["Marktwert"].replace("-", np.nan)
```

```python
bundesliga.isna().sum()
```

```python
import numpy as np

premier_league["Ablöse"] = premier_league["Ablöse"].replace({
    "ablösefrei": "0",
    "?": np.nan,
    "-": np.nan
})

premier_league["Marktwert"] = premier_league["Marktwert"].replace("-", np.nan)
```

```python
premier_league.isna().sum()
```
